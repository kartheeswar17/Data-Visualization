```python
file_path="/content/drive/MyDrive/AAPL.csv"
```

```python
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns

df=pd.read_csv(file_path)

df.head()
```

```python
plt.title('OHLC prices')
plt.xlabel('Date')
plt.ylabel('Price')
plt.legend()
plt.show()
```

```python
plt.figure(figsize=(14, 5))
sns.lineplot(data=df, x=df.index, y='Volume', color='orange')
plt.title('Trading Volume Over Time')
plt.xlabel('Date')
plt.ylabel('Volume')
plt.grid(True)
plt.show()
```
