```python
categorical_cols = {
    "gender",
    "race/ethnicity",
    "parental level of education",
    "lunch",
    "test preparation course"
}
```

```python
for col in categorical_cols:
  df[col] = df[col].str.strip().str.title()
print(df[list(categorical_cols)].head())
df['average_score'] = (
    df['math score'] +
    df['reading score'] +
    df['writing score']
) / 3
df['total_score'] = (
    df['math score'] +
    df['reading score'] +
    df['writing score']
)
```

```python
df.to_csv('/content/StudentsPerformance_Preprocessed.csv', index=False)
```
